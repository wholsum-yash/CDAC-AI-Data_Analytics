```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import sys

print(sys.executable)

# REFER DOCUMENTATION FOR HELP ON matplotlib
```

# Basic Graph

```python
x = [0,1,2,3,4]
y = [0,2,4,6,8]

# resize the graph
plt.figure(figsize=(5,3), dpi = 300)

# shorthand notation
# fmt = [color][marker][line] => mind the position, its important

# line 1
plt.plot(x, y, 'b^--', label = '2x')

# line 2
x2 = np.arange(0,4.5,0.5)
plt.plot(x2[:5], x2[:5]**2, color = 'orange', label = 'x^2')
plt.plot(x2[4:], x2[4:]**2, linestyle =  '--', color = 'red')

plt.title('Basic Graph')

plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.xticks([0,1,2,3])
plt.yticks([0,2,4,6])

plt.legend()

# plt.savefig('name_of_image.png', dpi = 300) saving the graph

plt.show()
```

# Bar Charts

```python
labels = ['A','B','C',]
values = [2,7,5]

bars = plt.bar(labels, values)

plt.figure(figsize=(9,5), dpi = 300)

# bars[0].set_hatch('/')
# bars[1].set_hatch('O')
# bars[2].set_hatch('*')

patterns = ['//', 'O', '*']
for bar in bars:
    bar.set_hatch(patterns.pop())
plt.show()
```

# Real World Examples

# Line Graph - `gas_prices.csv`

```python
gas = pd.read_csv('gas_prices.csv')
print(gas)

plt.figure(figsize=(9,5), dpi = 300)

# plt.plot(gas.Year, gas.USA) => not good practice

# plt.plot(gas['Year'], gas['Canada'], 'r.-',label = 'Canada')
# plt.plot(gas['Year'], gas['USA'], 'b.-',label = 'USA')
# plt.plot(gas['Year'], gas['South Korea'], 'g.-', label = 'South Korea')
# plt.plot(gas['Year'], gas['Australia'], 'y.-', label = 'Australia')
# plt.plot(gas['Year'], gas['Germany'], 'c.-', label = 'Germany')

# to quickly plot all the countries
for country in gas:
    if country != 'Year':
        plt.plot(gas['Year'], gas[country], marker = '.', label = country)

plt.title('Gas Prices across Countries', fontsize = 20)
plt.xlabel('Year')
plt.ylabel('Price')

plt.xticks(gas['Year'][::3]) # [::3] taking every 3rd year

# plt.legend()

# to push legend outside the graph
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

# `fifa_data.csv`

## - Histogram
## - Pie Chart
## - Box Plot

```python
fifa = pd.read_csv('fifa_data.csv')
print(fifa.head())

# histogram
bins = [40,50,60,70,80,90,100]

plt.title('Fifa Data')
plt.hist(fifa['Overall'], bins = bins)
plt.xticks(bins)
plt.xlabel('Overall of Fifa Player')
plt.ylabel('Volume of Fifa Player at a rating')
plt.show()
```

```python
# pie chart

left = fifa.loc[fifa['Preferred Foot'] == 'Left'].shape[0]
right = fifa.loc[fifa['Preferred Foot'] == 'Right'].shape[0]

plt.figure(figsize=(9,5), dpi = 300)
plt.pie([left, right], labels = ['left footed', 'right footed'], autopct='%1.1f%%')
plt.title('Preferred Foot of players in Fifa')
plt.show()

# pie chart 2

fifa['Weight'] = [int (x.strip('lbs')) if type(x) == str else x for x in fifa['Weight']]

light  = fifa.loc[fifa['Weight'] < 125].shape[0]
light_medium  = fifa.loc[(fifa['Weight'] >= 125) & (fifa['Weight'] < 150)].shape[0]
medium  = fifa.loc[(fifa['Weight'] >= 150) & (fifa['Weight'] < 175)].shape[0]
medium_heavy  = fifa.loc[(fifa['Weight'] >= 175) & (fifa['Weight'] < 200)].shape[0]
heavy  = fifa.loc[(fifa['Weight'] > 200)].shape[0]

weights = [light, light_medium, medium, medium_heavy, heavy]
label = ['Under 125', '125-150', '150-175,', '175-200', 'Over 200']
explode = (0.3, 0, 0, 0, 0.3)

plt.style.use('ggplot')
plt.figure(figsize=(9,5), dpi = 300)
plt.pie(weights, labels = label, autopct='%1.1f%%', pctdistance = 0.8, explode = explode)
plt.title('Player split by Weight in lbs')
plt.show()
```

```python
# box plot
plt.style.use('default')

barca = fifa.loc[fifa['Club'] == 'FC Barcelona']['Overall']
madrid = fifa.loc[fifa['Club'] == 'Real Madrid']['Overall']

labels = ['FC Barcelona', 'Real Madrid']

plt.boxplot([barca, madrid], labels = lables)
plt.show()
```
