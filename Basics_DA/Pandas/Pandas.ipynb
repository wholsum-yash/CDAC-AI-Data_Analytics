# Loading and Accessing Data from a `.csv` file

```python
# loading
coffee = pd.read_csv('coffee.csv')
results_p = pd.read_parquet('results.parquet')
bios = pd.read_csv('bios.csv')


# accessing
print(coffee.sample(10, random_state = 1)) # to make data deterministic we use random_state
print('-'*60)
# loc => filter dataframes by rows and columns and iloc 
print(coffee.loc[5:13]) 
print('-'*60)
print(coffee.iloc[5:13,[0,2]]) # uses only index values
print('-'*60)
```

# Adding and Removing Columns

```python
coffee_new = coffee.copy()
coffee_new['price'] = 4.99
coffee_new['new_price'] = np.where(coffee_new['Coffee Type'] == 'Espressor', 3.99, 5.99)

coffee_new.drop(columns=['price'], inplace = True)
# print(coffee_new)
# print('-'*60)

coffee_new['revenue'] = coffee_new['Units Sold'] * coffee_new['new_price']
# print(coffee_new['revenue'])
# print('-'*60)


coffee_new.rename(columns = {'new_price': 'price'}, inplace = True)
# print(coffee_new)
# print('-'*60)

bios_new = bios.copy()

print(bios_new.head())
print('-'*60)
bios_new['first name'] = bios_new['name'].str.split(" ").str[0]
display(bios_new)

bios_new['born_datetime'] = pd.to_datetime(bios_new['born_date']) # errors and format parameter
display(bios_new)
print(bios_new.info) 
```

# Merging and Concatenating Data

```python
bios.head()
```

```python
noc = pd.read_csv('noc_regions.csv')
bios = pd.read_csv('bios.csv')
results = pd.read_csv('results.csv')

# merging
bios_new2 = pd.merge(bios, noc, left_on = 'born_country', right_on = 'NOC', how = 'left')
bios_new2.rename(columns = { 'region': 'born_country_full'}, inplace = True)
bios_new2[bios_new2['NOC_x'] != bios_new2['born_country_full']][['name', 'NOC_x', 'born_country_full']]

usa = bios_new2[bios_new2['born_country'] == 'USA'].copy()
gbr = bios_new2[bios_new2['born_country'] == 'GBR'].copy()

new_df = pd.concat([usa,gbr])

# display(new_df.head())
# display(new_df.tail())

combined_df = pd.merge(results, bios, on = 'athlete_id', how = 'left')
# combined_df.head()
```

# Handling Null Values

```python
coffee.loc[[2,3], 'Units Sold'] = np.nan
coffee.isna().sum()

coffee_new2 = coffee.copy()
coffee_new2.fillna(coffee_new2['Units Sold'].interpolate()) 
# interpolate reads patterns in the data and continues that pattern in the data
```

# Aggregate Data

```python
bios_new2[bios_new2['born_country'] == 'USA']['born_region'].value_counts()

# cause coffee_new has price parameter
coffee_new.groupby(['Coffee Type']).agg({'Units Sold': 'sum', 'price': 'mean',})

pivot = coffee_new.pivot(columns = 'Coffee Type', index = 'Day', values = 'revenue')

pivot.loc['Monday', 'Latte']
pivot.sum()
pivot.sum(axis = 1)

bios_new['born_date'] = pd.to_datetime(bios['born_date'])
bios_new.groupby(bios_new['born_date'].dt.year)['name'].count().reset_index().sort_values('name', ascending = False)
```
